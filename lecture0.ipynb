In this short demo we compute the **NDVI** (Normalized Difference Vegetation Index),
one of the most widely used quantities in optical Earth Observation:

$$\mathrm{NDVI} = \frac{NIR - Red}{NIR + Red}$$

Healthy vegetation reflects strongly in near-infrared (NIR) and absorbs red light,
so NDVI is high (toward +1) over forests and crops, near 0 over bare soil, and negative over water.

```python
import numpy as np
import matplotlib.pyplot as plt
print("Environment OK — numpy", np.__version__)
```

## A toy scene

Before touching real satellite data, let's build a tiny artificial "scene":
a 100×100 image that is water on the left, bare soil in the middle, and forest on the right.

```python
rng = np.random.default_rng(42)
shape = (100, 100)

red = np.empty(shape)
nir = np.empty(shape)

# water (left third): low red, very low NIR
red[:, :33] = 0.05 + 0.01 * rng.random((100, 33))
nir[:, :33] = 0.02 + 0.01 * rng.random((100, 33))

# bare soil (middle third): moderate red and NIR
red[:, 33:66] = 0.25 + 0.05 * rng.random((100, 33))
nir[:, 33:66] = 0.30 + 0.05 * rng.random((100, 33))

# forest (right third): low red, high NIR
red[:, 66:] = 0.04 + 0.02 * rng.random((100, 34))
nir[:, 66:] = 0.45 + 0.10 * rng.random((100, 34))

ndvi = (nir - red) / (nir + red)

fig, ax = plt.subplots(1, 3, figsize=(12, 4))
for a, img, title in zip(ax, [red, nir, ndvi], ["Red", "NIR", "NDVI"]):
    im = a.imshow(img, cmap="RdYlGn" if title == "NDVI" else "gray",
                  vmin=-1 if title == "NDVI" else 0,
                  vmax=1 if title == "NDVI" else 0.6)
    a.set_title(title); a.axis("off")
    plt.colorbar(im, ax=a, shrink=0.7)
plt.tight_layout()
plt.show()
```

**Reading the result:** water shows strongly negative NDVI, soil near zero, forest strongly positive —
exactly the pattern the formula predicts. In the exercises we will compute the same index from real
Sentinel-2 measurements.
