## A first plot

Satellites measure reflected sunlight in different wavelengths. The curve below is a made-up but typical spectrum of green vegetation: low reflectance in the visible, a jump in the near-infrared. You will see real ones in Exercise 3.

```python
import numpy as np
import matplotlib.pyplot as plt

wavelength = np.array([450, 500, 550, 600, 650, 700, 750, 800, 850, 900])   # nanometres
reflectance = np.array([0.04, 0.05, 0.08, 0.06, 0.04, 0.10, 0.35, 0.45, 0.47, 0.48])

plt.figure(figsize=(6, 3.5))
plt.plot(wavelength, reflectance, marker="o")
plt.xlabel("Wavelength (nm)")
plt.ylabel("Reflectance")
plt.title("Typical spectrum of green vegetation")
plt.grid(alpha=0.3)
plt.show()
```
