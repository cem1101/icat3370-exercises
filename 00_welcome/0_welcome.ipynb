## A first plot. Why muddy water looks brown

A satellite measures how much sunlight the surface reflects at different wavelengths. Clear water reflects very little and less and less towards the red. Water carrying sediment reflects more, especially in the red. That difference is what the whole course is built on.

The two curves below are made up but typical.

```python
import numpy as np
import matplotlib.pyplot as plt

wavelength   = np.array([443, 490, 560, 665, 705, 740, 783, 842])   # Sentinel-2 band centres, nm
clear_water  = np.array([0.030, 0.028, 0.022, 0.010, 0.007, 0.004, 0.003, 0.002])
turbid_water = np.array([0.040, 0.055, 0.080, 0.075, 0.060, 0.040, 0.030, 0.020])

plt.figure(figsize=(6.5, 3.8))
plt.plot(wavelength, clear_water,  marker="o", label="clear water")
plt.plot(wavelength, turbid_water, marker="o", label="turbid (muddy) water")
plt.axvspan(600, 700, color="red", alpha=0.08)      # the red band region
plt.xlabel("Wavelength (nm)"); plt.ylabel("Reflectance")
plt.title("Typical reflectance of clear vs turbid water")
plt.legend(); plt.grid(alpha=0.3); plt.show()
```

Look at the red region (shaded). Muddy water reflects several times more than clear water there. Comparing the red band with the green band is the simplest way to map muddiness and you will do exactly that in Exercise 3.
